## **Modular Calculator**

## **What functions did you create in your calculator program?**
- the functions i created in my programes is add(a, b), subtract(a, b):, divide(a, b):, and mul(a, b):.

## **What parameters did your functions use?**

## **What arguments were passed when the functions were called?**
-the arguments passed during the functions were called was the inputted values by the user.
## **How did your program use the returned value?**

## **Why is it better to divide the calculator program into functions instead of writing everything in one long block of code?**
- putting it in one singular block of code makesthe code hard to understand, meanwhile if you split it up youll make it simpler.

```python
choice = input("please enter the operation you want to use, 1 for addition, 2 for subtraction, 3 for division, 4 for multiplication:")
# addition

def add(a, b):
  a = int(input("enter a first number: "))
  b = int(input("enter a second number: "))
  sum = a + b
  return sum

  print("The sum of the two numbers is:", add(a, b))

# subtraction

def subtract(a, b):
  a = int(input("enter a first number: "))
  b = int(input("enter a second number: "))
  difference = a - b
  return difference

  print("The difference of the two numbers is:", subtract(a, b))

# division

def divide(a, b):
  a = int(input("enter a first number: "))
  b = int(input("enter a second number: "))
  quotient = a / b
  return quotient

  print("The quotient of the two numbers is:", divide(a, b))

# multiplication

def mul(a, b):
  a = int(input("enter a first number: "))
  b = int(input("enter a second number: "))
  product = a * b
  return product

  print("The product of the two numbers is:", mul(a, b))


if choice == 1:
 add(a, b)

elif choice == 2:
 subtract(a, b)

elif choice == 3:
 divide(a, b)

elif choice == 4:
 mul(a, b)






```
