## **Modular Calculator**

## **What functions did you create in your calculator program?**
- the functions i created in my programes is add(a, b), subtract(a, b):, divide(a, b):, and mul(a, b):.

## **What parameters did your functions use?**
- my function used the parameters a and b, they are placeholders for the two digits the user is going to input.

## **What arguments were passed when the functions were called?**
-the arguments passed during the functions were called was the inputted values by the user.

## **How did your program use the returned value?**
- my program used the return value as a variable because the return function made the value usable for the program itself.

## **Why is it better to divide the calculator program into functions instead of writing everything in one long block of code?**
- putting it in one singular block of code makesthe code hard to understand, meanwhile if you split it up youll make it simpler.

```python
# Get the choice from the user (converted to an integer right away)
choice = int(input("Please enter the operation you want to use (1 for addition, 2 for subtraction, 3 for division, 4 for multiplication): "))

# Function definitions
def add():
    a = int(input("Enter a first number: "))
    b = int(input("Enter a second number: "))
    return a + b

def subtract():
    a = int(input("Enter a first number: "))
    b = int(input("Enter a second number:"))
    return a - b

def divide():
    a = int(input("Enter a first number: "))
    b = int(input("Enter a second number: "))
    if b == 0:
        return "Error! Division by zero."
    return a / b

def mul():
    a = int(input("Enter a first number: "))
    b = int(input("Enter a second number: "))
    return a * b

# Conditional block to handle the user's choice and print the result

if choice == 1:
    result = add()
    print("The sum of the two numbers is:", result)

elif choice == 2:
    result = subtract()
    print("The difference of the two numbers is:", result)

elif choice == 3:
    result = divide()
    print("The quotient of the two numbers is:", result)

elif choice == 4:
    result = mul()
    print("The product of the two numbers is:", result)

else:
    print("Invalid choice! Please restart the program and enter a number from 1 to 4.")
```
